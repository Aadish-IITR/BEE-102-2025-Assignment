```python
with open('multiline_input.fasta', 'r') as infile, open('new_sequence.fasta', 'w') as outfile:
    seq = ''
    for line in infile:
        line = line.strip()
        if line[0] == '>':
            if seq:
                outfile.write(seq + '\n')
            outfile.write(line + '\n')
            seq = ''
        else:
            seq += line
    if seq:
        outfile.write(seq + '\n')
```
